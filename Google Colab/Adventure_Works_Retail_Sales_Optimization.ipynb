```python
clean_tables = {
    "DimCustomer": dim_customer_clean,
    "DimEmployee": dim_employee_clean,
    "DimGeography": dim_geography_clean,
    "DimProduct": dim_product_clean,
    "DimReseller": dim_reseller_clean,
    "DimSalesTerritory": dim_sales_territory_clean,
    "FactInternetSales": fact_internet_sales_clean,
    "FactResellerSales": fact_reseller_sales_clean
}

for name, df in clean_tables.items():
    print("=" * 80)
    print(name)
    print("=" * 80)
    print("Shape:", df.shape)
    print("Columns:")
    print(list(df.columns))
    print()
```

```python
# Standardize integer columns
integer_columns = {
    "DimCustomer": ["CustomerKey", "GeographyKey", "YearlyIncome", "HouseOwnerFlag"],
    "DimEmployee": ["EmployeeKey", "SalesTerritoryKey"],
    "DimGeography": ["GeographyKey", "SalesTerritoryKey"],
    "DimProduct": ["ProductKey", "ProductSubcategoryKey"],
    "DimReseller": ["ResellerKey", "GeographyKey"],
    "DimSalesTerritory": ["SalesTerritoryKey"],
    "FactInternetSales": ["ProductKey", "CustomerKey", "SalesTerritoryKey", "SalesOrderLineNumber"],
    "FactResellerSales": ["ProductKey", "ResellerKey", "EmployeeKey", "SalesTerritoryKey", "SalesOrderLineNumber"]
}

for table_name, columns in integer_columns.items():
    df = clean_tables[table_name]

    for col in columns:
        if col in df.columns:
            # Nullable integer type is used where necessary
            df[col] = pd.to_numeric(df[col], errors="coerce").astype("Int64")

# Standardize date columns
date_columns = {
    "DimCustomer": ["BirthDate", "FirstPurchaseDate"],
    "DimEmployee": ["HireDate", "BirthDate"],
    "FactInternetSales": ["OrderDate", "DueDate", "ShipDate"],
    "FactResellerSales": ["OrderDate", "DueDate", "ShipDate"]
}

for table_name, columns in date_columns.items():
    df = clean_tables[table_name]

    for col in columns:
        if col in df.columns:
            df[col] = pd.to_datetime(df[col], errors="coerce")

print("Data types standardized successfully.")

print("\nFactInternetSales date types:")
print(fact_internet_sales_clean[["OrderDate", "DueDate", "ShipDate"]].dtypes)

print("\nFactResellerSales date types:")
print(fact_reseller_sales_clean[["OrderDate", "DueDate", "ShipDate"]].dtypes)
```

```python
# Determine the reference date from both sales channels
internet_max_date = fact_internet_sales_clean["OrderDate"].max()
reseller_max_date = fact_reseller_sales_clean["OrderDate"].max()

reference_date = max(internet_max_date, reseller_max_date)

# Calculate age using a vectorized method
dim_customer_clean["Age"] = (
    reference_date.year
    - dim_customer_clean["BirthDate"].dt.year
    - (
        (
            dim_customer_clean["BirthDate"].dt.month > reference_date.month
        )
        |
        (
            (dim_customer_clean["BirthDate"].dt.month == reference_date.month)
            & (dim_customer_clean["BirthDate"].dt.day > reference_date.day)
        )
    ).astype(int)
)

# Create age groups for marketing analysis
def assign_age_group(age):
    if age < 25:
        return "<25"
    elif age < 30:
        return "25-30"
    elif age < 35:
        return "30-35"
    elif age < 40:
        return "35-40"
    elif age < 45:
        return "40-45"
    elif age < 50:
        return "45-50"
    else:
        return "50+"

dim_customer_clean["AgeGroup"] = dim_customer_clean["Age"].apply(assign_age_group)

print("Reference Date:", reference_date.date())

print("\nAge statistics:")
print(dim_customer_clean["Age"].describe())

print("\nCustomer count by Age Group:")
print(
    dim_customer_clean["AgeGroup"]
    .value_counts()
    .sort_index()
)

print("\nAge columns added successfully.")

print(
    dim_customer_clean[
        ["CustomerKey", "BirthDate", "Age", "AgeGroup"]
    ].head(10)
)
```

```python
# Get the complete sales date range from both fact tables
min_date = min(
    fact_internet_sales_clean["OrderDate"].min(),
    fact_reseller_sales_clean["OrderDate"].min()
)

max_date = max(
    fact_internet_sales_clean["OrderDate"].max(),
    fact_reseller_sales_clean["OrderDate"].max()
)

# Create continuous date range
date_range = pd.date_range(
    start=min_date,
    end=max_date,
    freq="D"
)

# Create Date Dimension
dim_date = pd.DataFrame({
    "Date": date_range
})

# Date attributes
dim_date["Year"] = dim_date["Date"].dt.year
dim_date["Quarter"] = "Q" + dim_date["Date"].dt.quarter.astype(str)
dim_date["MonthNumber"] = dim_date["Date"].dt.month
dim_date["MonthName"] = dim_date["Date"].dt.month_name()
dim_date["YearMonth"] = dim_date["Date"].dt.strftime("%Y-%m")
dim_date["Day"] = dim_date["Date"].dt.day
dim_date["DayName"] = dim_date["Date"].dt.day_name()
dim_date["WeekNumber"] = dim_date["Date"].dt.isocalendar().week.astype(int)

print("Date Dimension created successfully.")
print("Date Range:", dim_date["Date"].min().date(), "to", dim_date["Date"].max().date())
print("Total Dates:", len(dim_date))

print("\nColumns:")
print(list(dim_date.columns))

print("\nSample records:")
print(dim_date.head())

print("\nYear-wise date count:")
print(dim_date.groupby("Year")["Date"].count())
```

```python
# Add sales channel to distinguish the two sales sources

fact_internet_sales_clean["SalesChannel"] = "Internet"
fact_reseller_sales_clean["SalesChannel"] = "Reseller"

print("Sales Channel added successfully.")

print("\nInternet Sales:")
print(fact_internet_sales_clean["SalesChannel"].value_counts())

print("\nReseller Sales:")
print(fact_reseller_sales_clean["SalesChannel"].value_counts())

print("\nUpdated FactInternetSales shape:", fact_internet_sales_clean.shape)
print("Updated FactResellerSales shape:", fact_reseller_sales_clean.shape)
```

```python
# Select common columns from both sales fact tables
common_columns = [
    "ProductKey",
    "SalesTerritoryKey",
    "SalesOrderNumber",
    "SalesOrderLineNumber",
    "DiscountAmount",
    "TotalProductCost",
    "SalesAmount",
    "Freight",
    "CarrierTrackingNumber",
    "OrderDate",
    "DueDate",
    "ShipDate",
    "SalesChannel"
]

# Add CustomerKey for Internet Sales
internet_combined = fact_internet_sales_clean[
    common_columns[:]
].copy()

internet_combined["CustomerKey"] = fact_internet_sales_clean["CustomerKey"]
internet_combined["ResellerKey"] = pd.NA
internet_combined["EmployeeKey"] = pd.NA

# Add ResellerKey and EmployeeKey for Reseller Sales
reseller_combined = fact_reseller_sales_clean[
    common_columns[:]
].copy()

reseller_combined["CustomerKey"] = pd.NA
reseller_combined["ResellerKey"] = fact_reseller_sales_clean["ResellerKey"]
reseller_combined["EmployeeKey"] = fact_reseller_sales_clean["EmployeeKey"]

# Reorder columns
combined_columns = [
    "ProductKey",
    "CustomerKey",
    "ResellerKey",
    "EmployeeKey",
    "SalesTerritoryKey",
    "SalesOrderNumber",
    "SalesOrderLineNumber",
    "DiscountAmount",
    "TotalProductCost",
    "SalesAmount",
    "Freight",
    "CarrierTrackingNumber",
    "OrderDate",
    "DueDate",
    "ShipDate",
    "SalesChannel"
]

# Combine both channels
fact_sales = pd.concat(
    [
        internet_combined[combined_columns],
        reseller_combined[combined_columns]
    ],
    ignore_index=True
)

print("Combined Fact Sales created successfully.")
print("Shape:", fact_sales.shape)

print("\nSales Channel distribution:")
print(fact_sales["SalesChannel"].value_counts())

print("\nColumns:")
print(list(fact_sales.columns))

print("\nSample records:")
print(fact_sales.head())
```

```python
# Merge territory information with the combined sales fact
territory_sales = fact_sales.merge(
    dim_sales_territory_clean[
        [
            "SalesTerritoryKey",
            "SalesTerritoryRegion",
            "SalesTerritoryCountry",
            "SalesTerritoryGroup"
        ]
    ],
    on="SalesTerritoryKey",
    how="left"
)

# Calculate territory-level performance
territory_summary = (
    territory_sales
    .groupby(
        [
            "SalesTerritoryKey",
            "SalesTerritoryRegion",
            "SalesTerritoryCountry",
            "SalesTerritoryGroup"
        ],
        dropna=False
    )
    .agg(
        TotalSales=("SalesAmount", "sum"),
        TotalOrders=("SalesOrderNumber", "nunique"),
        SalesLines=("SalesOrderLineNumber", "count")
    )
    .reset_index()
)

# Calculate Average Order Value
territory_summary["AverageOrderValue"] = (
    territory_summary["TotalSales"]
    / territory_summary["TotalOrders"]
)

# Calculate sales contribution
territory_summary["SalesShare"] = (
    territory_summary["TotalSales"]
    / total_sales
    * 100
)

# Sort by total sales
territory_summary = territory_summary.sort_values(
    "TotalSales",
    ascending=False
).reset_index(drop=True)

print("=" * 80)
print("SALES TERRITORY PERFORMANCE")
print("=" * 80)

print(territory_summary.to_string(index=False))

print("\n" + "=" * 80)
print("LOWEST SALES TERRITORIES")
print("=" * 80)

print(
    territory_summary[
        [
            "SalesTerritoryRegion",
            "TotalSales",
            "TotalOrders",
            "AverageOrderValue",
            "SalesShare"
        ]
    ]
    .sort_values("TotalSales")
    .head(5)
    .to_string(index=False)
)
```

```python
# Calculate median benchmarks
sales_median = territory_summary["TotalSales"].median()
aov_median = territory_summary["AverageOrderValue"].median()

# Classify territory performance
def classify_territory(row):
    if (
        row["TotalSales"] < sales_median
        and row["AverageOrderValue"] < aov_median
    ):
        return "Underperforming"
    elif (
        row["TotalSales"] >= sales_median
        and row["AverageOrderValue"] >= aov_median
    ):
        return "High Performing"
    elif row["TotalSales"] >= sales_median:
        return "High Sales / Low AOV"
    else:
        return "Low Sales / High AOV"

territory_summary["PerformanceCategory"] = territory_summary.apply(
    classify_territory,
    axis=1
)

print("=" * 80)
print("TERRITORY PERFORMANCE BENCHMARKS")
print("=" * 80)

print(f"\nMedian Total Sales: ${sales_median:,.2f}")
print(f"Median Average Order Value: ${aov_median:,.2f}")

print("\n" + "=" * 80)
print("TERRITORY CLASSIFICATION")
print("=" * 80)

print(
    territory_summary[
        [
            "SalesTerritoryRegion",
            "TotalSales",
            "AverageOrderValue",
            "SalesShare",
            "PerformanceCategory"
        ]
    ].to_string(index=False)
)

print("\n" + "=" * 80)
print("UNDERPERFORMING TERRITORIES")
print("=" * 80)

underperforming = territory_summary[
    territory_summary["PerformanceCategory"] == "Underperforming"
]

print(
    underperforming[
        [
            "SalesTerritoryRegion",
            "TotalSales",
            "AverageOrderValue",
            "SalesShare",
            "PerformanceCategory"
        ]
    ].sort_values("TotalSales")
    .to_string(index=False)
)
```

```python
# Create annual income bands
def assign_income_group(income):
    if income < 30000:
        return "<30K"
    elif income < 50000:
        return "30K-50K"
    elif income < 75000:
        return "50K-75K"
    elif income < 100000:
        return "75K-100K"
    else:
        return "100K+"

customer_summary["IncomeGroup"] = customer_summary["YearlyIncome"].apply(
    assign_income_group
)

# Aggregate sales by income group
income_summary = (
    customer_summary
    .groupby("IncomeGroup", dropna=False)
    .agg(
        Customers=("CustomerKey", "nunique"),
        TotalSales=("TotalSales", "sum"),
        AverageSalesPerCustomer=("TotalSales", "mean"),
        AverageIncome=("YearlyIncome", "mean")
    )
    .reset_index()
)

# Keep income groups in logical order
income_order = [
    "<30K",
    "30K-50K",
    "50K-75K",
    "75K-100K",
    "100K+"
]

income_summary["IncomeGroup"] = pd.Categorical(
    income_summary["IncomeGroup"],
    categories=income_order,
    ordered=True
)

income_summary = income_summary.sort_values("IncomeGroup")

print("=" * 80)
print("CUSTOMER SEGMENTATION BY ANNUAL INCOME")
print("=" * 80)

print(
    income_summary.to_string(index=False)
)

print("\n" + "=" * 80)
print("HIGHEST VALUE INCOME SEGMENTS")
print("=" * 80)

print(
    income_summary
    .sort_values("AverageSalesPerCustomer", ascending=False)
    .head(3)
    .to_string(index=False)
)
```

```python
# Sales channel performance by year
channel_yearly = (
    fact_sales
    .assign(Year=fact_sales["OrderDate"].dt.year)
    .groupby(["Year", "SalesChannel"])
    .agg(
        TotalSales=("SalesAmount", "sum"),
        TotalOrders=("SalesOrderNumber", "nunique"),
        GrossProfit=("SalesAmount", lambda x: x.sum()),
        TotalCost=("TotalProductCost", "sum")
    )
    .reset_index()
)

# Calculate gross profit and AOV
channel_yearly["GrossProfit"] = (
    channel_yearly["TotalSales"]
    - channel_yearly["TotalCost"]
)

channel_yearly["ProfitMargin"] = (
    channel_yearly["GrossProfit"]
    / channel_yearly["TotalSales"]
    * 100
)

channel_yearly["AverageOrderValue"] = (
    channel_yearly["TotalSales"]
    / channel_yearly["TotalOrders"]
)

print("=" * 80)
print("SALES CHANNEL PERFORMANCE BY YEAR")
print("=" * 80)

print(
    channel_yearly.to_string(index=False)
)

print("\n" + "=" * 80)
print("OVERALL SALES CHANNEL PERFORMANCE")
print("=" * 80)

channel_overall = (
    fact_sales
    .groupby("SalesChannel")
    .agg(
        TotalSales=("SalesAmount", "sum"),
        TotalOrders=("SalesOrderNumber", "nunique"),
        TotalCost=("TotalProductCost", "sum")
    )
    .reset_index()
)

channel_overall["GrossProfit"] = (
    channel_overall["TotalSales"]
    - channel_overall["TotalCost"]
)

channel_overall["ProfitMargin"] = (
    channel_overall["GrossProfit"]
    / channel_overall["TotalSales"]
    * 100
)

channel_overall["AverageOrderValue"] = (
    channel_overall["TotalSales"]
    / channel_overall["TotalOrders"]
)

channel_overall["SalesShare"] = (
    channel_overall["TotalSales"]
    / total_sales
    * 100
)

print(
    channel_overall.to_string(index=False)
)
```
